This data is from 2025 surveys of the Floating Kelp Kayak Site of Possession Point, monitored by the Marine Resource Committie of Island County. This is just one of 5 sites where such data is taken, however, this is the only site that has a full year of consistent data.

It consists of data collected via kayak using a CTD (conductivity, temperature, and dissolved solids) monitor which can take Temperature, Salinity, and Conductivity measurements from the seasurface down to 35ft.

In short, this data is taken to search for three distict oceanonagrphic points that dilineate stratification. The halocline, a rapid change in salinity and water density due to seperation of different chemical makeups of water from land and sea. The thermocline, the rapid change in temperature and water density due to differences in mixing from upwelling water in the pacific and snow melt/rain runoff from the land.

Together, they make an emergent seperation of water known as the pycnocline which can influence kelp reproduction, exposing spores that grow on the ocean surface and their release and settling on the ocean floor.

**The Variables in this Data are As Follows:**

*site* -{object} The Name of the Site

*date* -{object} Date of Survey

*data_point* -{object} 4 data points are taken during each survey, two along the
deep edge of the bed (DE1, DE2) and two along the shore (SE1, SE2)

*temp* -{float64} Temperature in Degrees Celcius

*salinity* -{float64} - Salinity in Parts Per Thousand

*conductivity* -{float64} - Conductivity in Milisiemens

*depth* -{int64} - Depth in Feet, at each depth interval of 2 feet a Salinity, Conductivity and Temperature are typically taken.

```python
alt.Chart(ctd).mark_bar().encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('temp'), bin=True).title('Temperature (C)'),
    alt.Column('data_point')
)
```

This plot does show some minor stratification in the upper 4 feet through out the summer, and potentially some intersting patters from month to month. However, it needs some help to really bring out what the data is saying. It does seem that there is some peaking in termperature generally in august, however, the seperation across each data point makes that hard to truthfully judge.

```python
alt.Chart(ctd).mark_bar().encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('salinity'), bin=True).title('Salinity (ppt)'),
    alt.Column('data_point')
)
```

This shows a much stronger level of stratification from temperature and may be more interesting to pursue, however, temperature is known to have a stronger effect on kelp.

It is intersting to see the spike in overall salinity in the September surveys and a generally stronger stratification in the early months of the summer. However, again this is tough with the position seperation by each survey site.

```python
alt.Chart(ctd).mark_line(color='firebrick').encode(
    alt.Color('month(date)').title('Survey Month'),
    alt.Y('depth').title('depth').scale(reverse = True),
    alt.X('mean(temp)').title('Temperature (C)').scale(reverse = True),
)
```

This aggregation method by averaging temp across data points cleans up the line plot a great deal. The most intersting pattern to me here, is that the surveys that did not see generally deeper maximum depths saw, shallowe slope in their temperature stratification. If I were to continue to iterate this plot into my refined plots, I would make the colors more obvious wich month each line belongs to.

```python
alt.Chart(ctd).mark_line(color='firebrick').encode(
    alt.Color('month(date)').title('Survey Month'),
    alt.Y('depth').title('depth').scale(reverse = True),
    alt.X('mean(salinity)').title('Salinity (ppt)'),
)
```

Similar to temperature this aggregations allows us to see a great deal more. Similarly to the heat map plots as well, we can see that salinity has a much steaper stratification gradient than that of temperature. Opposed to temperature, however, the shallower surveys do still have steep gradients.

# **Refined Plots**

```python
alt.Chart(ctd).mark_bar().encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(temp)'), bin=True).title('Mean Temperature (C)'),
    alt.Column('data_point')
)
```

Starting simple, lets just see what mean of temperature can do to try to bring out the stratification.

In fact, this aggregation strategy further blurs the lines as it is averaging across just one reading. It seems to make altair change the color gradient, however, which is intersting in itself but does not really give us anything enlightening about the data.

```python
alt.Chart(ctd).mark_bar().encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(temp)'), bin=True).title('Mean Temperature (C)'),
)
```

This works much better because now we are averging accross data points in the survey which gives us a better picture of how the stratification may be in the entire bed. Howver, since strtification is a measure of the change over depth, let us find the difference.

```python
alt.Chart(ctd).mark_bar().transform_filter(
    'datum.temp != null',
).transform_window(
    surf_temp ='first_value(temp)',
    frame=[None, None],
    groupby=['date', 'data_point'],
).transform_calculate(
    thermo= 'datum.temp - datum.surf_temp',
).encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(thermo):Q')).title('Change Temperature from Surface')
)
```

This makes a much stronger showing of our change over depth by taking the difference from the surface temperature. This transformation window [found first here](https://altair-viz.github.io/altair-viz-v4/user_guide/transform/window.html) technique really gets to the hear of what we want. Now lets try with salinity.

```python
alt.Chart(ctd).mark_bar().transform_filter(
    'datum.salinity != null',
).transform_window(
    surf_sal ='first_value(salinity)',
    frame=[None, None],
    groupby=['date', 'data_point'],
).transform_calculate(
    halo= 'datum.salinity - datum.surf_sal',
).encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(halo):Q')).title('Change Salinity from Surface'),
)
```

As we saw in our exploratory and preliminary graphs, salinity has a much stronger gradient and thus these graphs tend to have turned out even better than temperature. Howver, since they both have influence on kelp reproduction it is important we do what we can to compare them, even if imperfect. I plan to put them side by side in the final graph.

```python
temp = alt.Chart(ctd).mark_bar().transform_filter(
    'datum.temp != null',
).transform_window(
    surf_temp ='first_value(temp)',
    frame=[None, None],
    groupby=['date', 'data_point'],
).transform_calculate(
    thermo= 'datum.temp - datum.surf_temp',
).encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(thermo):Q')).title('Change in Temperature').scale(reverse = True),
)

sal = alt.Chart(ctd).mark_bar().transform_filter(
    'datum.salinity != null',
).transform_window(
    surf_sal ='first_value(salinity)',
    frame=[None, None],
    groupby=['date', 'data_point'],
).transform_calculate(
    halo= 'datum.salinity - datum.surf_sal',
).encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(halo):Q')).title('Change in Salinity').scale(scheme = 'brownbluegreen'),
)

(temp | sal)
```

# **Final Plot**

```python
temp = alt.Chart(ctd).mark_bar().transform_filter(
    'datum.temp != null',
).transform_window(
    surf_temp ='first_value(temp)',
    frame=[None, None],
    groupby=['date', 'data_point'],
).transform_calculate(
    thermo= 'datum.temp - datum.surf_temp',
).encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(thermo):Q')).title('Change in Temperature').scale(reverse = True),
)

sal = alt.Chart(ctd).mark_bar().transform_filter(
    'datum.salinity != null',
).transform_window(
    surf_sal ='first_value(salinity)',
    frame=[None, None],
    groupby=['date', 'data_point'],
).transform_calculate(
    halo= 'datum.salinity - datum.surf_sal',
).encode(
    alt.X('month(date)').title('Survey Month'),
    alt.Y('depth').bin(step=2).title('Depth (ft)').scale(reverse = True),
    alt.Color(('mean(halo):Q')).title('Change in Salinity').scale(scheme = 'brownbluegreen', domainMid=0),
)

(temp | sal).resolve_scale(color='independent')
```

This plot depicts the change in thermocline and haloclines at Possession Point Floating Kelp Bed in 2025 from June through September. This information is important to inform decision making and health tracking of kelp reproduction and survival.

The data made for some intuitive encoding decisions, depth being on the y axis and reversed since we take measurments from the surface down, time, or month, on the y axis to compare seasonality, and the final quantitative value as the color saturation to represent relative speed of gradient change in salinity and or temperature.However, even with intuition on my side, we needed to show more clearly how gradient was changing, prompting me to recall our discussions of differences in class as an important transformation in our toolkit. In the end this too was semi-intuitive as the definition of the pynocline is best identified as a sudden change in density indicated by values such as temperature and salinity. Finally, I decided to create one last position encoding by placing thermocline and halocline graphs side by side as it can show the difference in the stratification between enviromental parameters and help inform pycnocline estimation.

We can see that there is a clear seperation between the top 4 feet of the water column and the bottom rest; especially in the months of June, August, and September.
